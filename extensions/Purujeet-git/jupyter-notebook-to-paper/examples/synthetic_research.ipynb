# Synthetic Research Study

## Research question

Does Model B outperform Model A on a synthetic classification task?

This notebook is synthetic and is designed to exercise metrics, statistics, figures, and equations.

```python
model_a_accuracy = 0.847
model_b_accuracy = 0.881
model_a_f1 = 0.826
model_b_f1 = 0.864
print(f"Model A accuracy: {model_a_accuracy:.3f}")
print(f"Model B accuracy: {model_b_accuracy:.3f}")
print(f"Model A F1: {model_a_f1:.3f}")
print(f"Model B F1: {model_b_f1:.3f}")
```

```python
# Example statistical result retained in the notebook output.p_value = 0.00012print(f"Paired t-test p-value: {p_value:.5f}")
```

```python
import matplotlib.pyplot as plt
model_names = ["Model A", "Model B"]
accuracy = [model_a_accuracy, model_b_accuracy]
plt.bar(model_names, accuracy)
plt.ylabel("Accuracy")
plt.title("Synthetic Model Accuracy Comparison")
plt.show()
```

## Statistical interpretation

Model B has a higher accuracy and F1 score than Model A in this synthetic evaluation. The paired comparison provides a reproducible statistical result for the paper pipeline.

The F1 score is defined as:

$$
F_1 = \frac{2PR}{P+R}
$$

where $P$ is precision and $R$ is recall.
